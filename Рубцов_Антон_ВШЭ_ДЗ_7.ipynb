```python
import requests

def get_first_five_posts():
    url = "https://jsonplaceholder.typicode.com/posts"

    # Отправляем GET-запрос к API
    response = requests.get(url)

    # Проверяем, что запрос прошел успешно
    if response.status_code == 200:
        posts = response.json()  # Преобразуем JSON-ответ в список словарей

        print("Первые 5 постов из JSONPlaceholder:\n" + "=" * 50)

        # Берем первые 5 постов с помощью среза [:5]
        for i, post in enumerate(posts[:5], start=1):
            print(f"Пост №{i} (ID: {post['id']})")
            print(f"Заголовок: {post['title']}")
            print(f"Тело поста:\n{post['body']}")
            print("-" * 50)
    else:
        print(f"Ошибка при запросе данных. Код ответа: {response.status_code}")

if __name__ == "__main__":
    get_first_five_posts()
```

```python
import requests

def get_weather(city_name, api_key):
    url = "https://api.openweathermap.org/data/2.5/weather"

    # Формируем параметры GET-запроса
    params = {
        "q": city_name,
        "appid": api_key,
        "units": "metric",  # Температура в градусах Цельсия (°C)
        "lang": "ru"        # Описание погоды на русском языке
    }

    # Отправляем GET-запрос с параметрами
    response = requests.get(url, params=params)

    if response.status_code == 200:
        data = response.json()

        # Извлекаем температуру и описание погоды из JSON-ответа
        temp = data["main"]["temp"]
        feels_like = data["main"]["feels_like"]
        description = data["weather"][0]["description"]
        city_display = data["name"]

        print(f"\nПогода в городе {city_display}:")
        print(f"Текущая температура: {temp}°C (ощущается как {feels_like}°C)")
        print(f"Описание погоды: {description.capitalize()}")

    elif response.status_code == 404:
        print(f"\nОшибка: Город «{city_name}» не найден. Проверьте правильность написания.")
    elif response.status_code == 401:
        print("\nОшибка 401: Неверный API-ключ или он еще не активировался.")
    else:
        print(f"\nОшибка при получении данных. Код ответа: {response.status_code}")

if __name__ == "__main__":
    API_KEY = "100a0bdb38da88ffc0062dd42fe21158"

    city = input("Введите название города: ").strip()
    if city:
        get_weather(city, API_KEY)
    else:
        print("Вы не ввели название города.")
```
